Список городов-милионников надо взять [отсюда](https://ru.wikipedia.org/wiki/%D0%93%D0%BE%D1%80%D0%BE%D0%B4%D0%B0-%D0%BC%D0%B8%D0%BB%D0%BB%D0%B8%D0%BE%D0%BD%D0%B5%D1%80%D1%8B_%D0%A0%D0%BE%D1%81%D1%81%D0%B8%D0%B8). 

Если возникнут трудности с этим задание посмотрите материалы модуля  PYTHON-17. Как получать данные из веб-источников и API. 

```python
# код для получения списка городов-милионников
import requests, re
from bs4 import BeautifulSoup

URL = f"https://api.wikimedia.org/core/v1/wikipedia/ru/page/Города-миллионеры_России/html"
HEADERS = {
    "User-Agent": "MillionCitiesFetcher/1.0 (+contact: your_email@example.com)"
}
# переменная для сохранения результата запроса
r = requests.get(URL, headers=HEADERS, timeout=20)
r.raise_for_status()
# структурируем информацию
soup = BeautifulSoup(r.text, "html.parser")
# основной список — первая таблица
table = soup.find("table")
# нам потребуется переменная для списка
rows = []
# разбираем текст и формируем список
for tr in table.find_all("tr")[1:]:
    tds = [td.get_text(strip=True) for td in tr.find_all("td")]
    if len(tds) < 3:
        continue
    if not tds[0].isdigit():
        continue
    city = tds[1]
    pop_2025 = int(re.sub(r"[^\d]", "", tds[2])) if tds[2] else 0
    if pop_2025 >= 1000:
        rows.append(city)

print(", ".join(rows))
```

```python
# текст запроса
query_5_7 = f'''
select
    ar.name,
    count(v.id) cnt
from
    vacancies as v
join
    employers as em on v.employer_id = em.id
join
    areas as ar on ar.id = v.area_id
where em.name like 'Яндекс'
group by 1
order by 2
'''
```

```python
# результат запроса отсортировать
df = pd.read_sql_query(query_5_7, connection)
# сортировка запроса
df = df[df.name.isin(rows)]

# строка для добавления в таблицу
now_df = pd.DataFrame([{'name': 'Total', 'cnt': df.cnt.sum()}])
df = pd.concat([df, now_df], ignore_index=True)
df
```

6. Напишите запрос, позволяющий вычислить, какую зарплату для DS в **среднем** указывают для каждого типа требуемого опыта (уникальное значение из поля *experience*). 

При решении задачи примите во внимание следующее:
1. Рассматриваем только вакансии, у которых заполнено хотя бы одно из двух полей с зарплатой.
2. Если заполнены оба поля с зарплатой, то считаем зарплату по каждой вакансии как сумму двух полей, делённую на 2. Если заполнено только одно из полей, то его и считаем зарплатой по вакансии.
3. Если в расчётах участвует null, в результате он тоже даст null (посмотрите, что возвращает запрос select 1 + null). Чтобы избежать этой ситуацию, мы воспользуемся функцией [coalesce](https://postgrespro.ru/docs/postgresql/9.5/functions-conditional#functions-coalesce-nvl-ifnull), которая заменит null на значение, которое мы передадим. Например, посмотрите, что возвращает запрос `select 1 + coalesce(null, 0)`

Выясните, на какую зарплату в среднем может рассчитывать дата-сайентист с опытом работы от 3 до 6 лет. Результат округлите до целого числа. 

```python
query_6_6 = f'''
select
    v.experience,
    round(avg(
            case
                when v.salary_from is not null and v.salary_to is not null then
                    (v.salary_from + v.salary_to) / 2
                when v.salary_from is not null then
                    v.salary_from
                when v.salary_to is not null then
                    v.salary_to
            end
    )) as avg_salary
    
from
    vacancies as v
where
    (
        lower(v.name) like '%data scientist%' or
        lower(v.name) like '%data science%' or
        lower(v.name) like '%исследователь данных%' or
        (
                    v.name like '%ML%' and
                    v.name not like '%HTML%'
        ) or
        lower(v.name) like '%machine learning%' or
        lower(v.name) like '%машинн%обучен%'
    )
group by v.experience
order by v.experience
'''
```

```python
df = pd.read_sql_query(query_6_6, connection)
df
```

```python
# Наиболее ценные навыки на рынке труда (зарплата > 200000)
query_6_7 = f'''
SELECT
    -- убираем пробелы из навыков
    TRIM(regexp_split_to_table(key_skills, E'\\t')) as skill,
    COUNT(*) as frequency,
    ROUND(AVG((salary_from + salary_to) / 2)::numeric, 0) as avg_salary
FROM vacancies
WHERE salary_from IS NOT NULL
  AND salary_to IS NOT NULL
  AND (salary_from + salary_to) / 2 > 200000
  AND key_skills IS NOT NULL
GROUP BY skill
ORDER BY frequency DESC
LIMIT 30;
'''
```
